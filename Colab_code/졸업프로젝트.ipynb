```
import json
import glob
import sqlite3
import os
import shutil
import zipfile
from tqdm import tqdm

# --- [단계 1] 경로 설정 ---
local_extract_path = "/content/temp_data"
local_db_path = "/content/LawData.db"
final_db_path = "/content/drive/MyDrive/졸프데이터/LawData.db"

# --- [단계 2] 내부 ZIP 파일들 전부 해제 ---
# .zip 혹은 .part0 등으로 끝나는 압축 파일들을 다 찾습니다.
internal_zips = glob.glob(f"{local_extract_path}/**/*.zip*", recursive=True)
print(f"📦 내부 압축 파일 {len(internal_zips)}개를 발견했습니다. 해제를 시작합니다.")

for zip_path in tqdm(internal_zips, desc="내부 압축 해제 중"):
    try:
        # .part0 같은 경우도 unzip 명령어로 강제 해제 시도
        !unzip -o -q "{zip_path}" -d "{local_extract_path}/final_json"
    except:
        continue

print("✅ 모든 내부 압축 해제 완료!")

# --- [단계 3] DB 초기화 ---
if os.path.exists(local_db_path):
    os.remove(local_db_path)

conn = sqlite3.connect(local_db_path)
cursor = conn.cursor()
cursor.execute("PRAGMA synchronous = OFF")
cursor.execute("PRAGMA journal_mode = WAL")

cursor.execute('''
    CREATE TABLE legal_data (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        case_name TEXT,
        case_num TEXT,
        court_name TEXT,
        content TEXT
    )
''')

# --- [단계 4] 진짜 JSON 찾기 ---
# 이제 풀린 final_json 폴더 내에서 모든 json을 찾습니다.
json_files = glob.glob(f"{local_extract_path}/**/*.json", recursive=True)
print(f"🚀 드디어 {len(json_files)}개의 JSON 파일을 찾았습니다!")

if len(json_files) == 0:
    print("❌ 아직도 JSON이 없습니다. 폴더 구조를 다시 확인해야 합니다.")
else:
    batch = []
    for file_path in tqdm(json_files, desc="DB 삽입 중"):
        try:
            with open(file_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
                # AI-Hub 데이터 특유의 구조(info, label) 확인
                info = data.get('info', {})
                label = data.get('label', {})

                output_content = label.get('output')
                if output_content:
                    batch.append((
                        info.get('caseName', '알 수 없음'),
                        info.get('caseNum', ''),
                        info.get('courtName', ''),
                        output_content
                    ))

            if len(batch) >= 5000:
                cursor.executemany('INSERT INTO legal_data (case_name, case_num, court_name, content) VALUES (?, ?, ?, ?)', batch)
                conn.commit()
                batch = []
        except:
            continue

    if batch:
        cursor.executemany('INSERT INTO legal_data (case_name, case_num, court_name, content) VALUES (?, ?, ?, ?)', batch)
        conn.commit()

conn.close()
shutil.copy(local_db_path, final_db_path)
print(f"✨ 진짜 완료! 최종 위치: {final_db_path}")
```

```
import sqlite3
import pandas as pd
from sentence_transformers import SentenceTransformer
import pickle
import os

# 1. 모델 불러오기 (한국어 성능이 좋은 모델 선택)
# 모바일(ObjectBox)에서도 쓸 수 있게 가벼운 모델을 사용합니다.
model = SentenceTransformer('snunlp/KR-SBERT-V40K-klueNLI-augSTS')

# 2. DB에서 데이터 읽어오기
conn = sqlite3.connect("/content/drive/MyDrive/졸프데이터/LawData.db")
# 일단 테스트로 1000개만 가져와 봅시다. (나중에 90000으로 바꾸면 됩니다)
df = pd.read_sql_query("SELECT id, content FROM legal_data LIMIT 1000", conn)
conn.close()

print(f" {len(df)}개의 데이터를 불러왔습니다.")

# 3. 임베딩 시작 (여기가 시간이 제일 많이 걸려요!)
print("🚀 임베딩 변환 시작... (GPU가 켜져 있으면 훨씬 빠릅니다)")
sentences = df['content'].tolist()
embeddings = model.encode(sentences, show_progress_bar=True, device='cuda') # GPU 사용

print("✨ 임베딩 완료!")

# 4. 결과 저장 (나중에 ObjectBox에 한꺼번에 넣기 위해 파일로 저장)
output_path = "/content/drive/MyDrive/졸프데이터/embeddings_test.pkl"
with open(output_path, 'wb') as f:
    pickle.dump({'ids': df['id'].tolist(), 'embeddings': embeddings}, f)

print(f" 임베딩 데이터가 저장되었습니다: {output_path}")
```

```
import pickle
import numpy as np
from sentence_transformers import SentenceTransformer, util
import sqlite3
import pandas as pd

# 1. 아까 만든 숫자와 모델 불러오기
with open("/content/drive/MyDrive/졸프데이터/embeddings_final.pkl", "rb") as f:
    data = pickle.load(f)

# 리스트를 계산하기 편한 numpy 배열로 변환
all_embeddings = np.array(data['embeddings'])
all_ids = data['ids']

# 2. 테스트용 질문 입력
user_question = "친구가 돈을 빌려가서 연락이 안 되는데 이거 처벌 가능한가요?"

# 3. 질문 임베딩
question_embedding = model.encode(user_question, device='cuda')

# 4. 유사도 계산(9만개 중 가장 비슷한 3개만 뽑아옴)
cos_scores = util.cos_sim(question_embedding, all_embeddings)[0]
top_results = np.argpartition(-cos_scores, range(3))[:3]

print(f"\n 질문: {user_question}")
print("-" * 50)

# 5. 결과 출력 (DB에서 실제 내용 가져오기)
conn = sqlite3.connect("/content/drive/MyDrive/졸프데이터/LawData.db")

for i, idx in enumerate(top_results):
    target_id = all_ids[idx]
    score = cos_scores[idx]

    # 해당 ID의 판례 내용 가져오기
    res = pd.read_sql_query(f"SELECT content FROM legal_data WHERE id='{target_id}'", conn)
    content = res['content'][0]

    print(f"[{i+1}] 유사도 점수: {score:.4f}")
    print(f" 판례 내용 요약: {content[:150]}...") # 너무 길면 잘라서 출력
    print("-" * 50)

conn.close()
```

```
import pandas as pd
import sqlite3
import pickle

# 1. 원본 데이터와 임베딩 데이터 다시 불러오기
conn = sqlite3.connect("/content/drive/MyDrive/졸프데이터/LawData.db")
df = pd.read_sql_query("SELECT id, content FROM legal_data", conn)
conn.close()

with open("/content/drive/MyDrive/졸프데이터/embeddings_final.pkl", "rb") as f:
    emb_data = pickle.load(f)

# 2. 데이터 합치기 (임베딩 숫자를 문자열로 변환하여 저장)
# 리스트 형태의 임베딩을 ','로 구분된 문자열로 바꿔야 CSV에 들어갑니다.
df['embedding'] = [",".join(map(str, e)) for e in emb_data['embeddings']]

# 3. CSV로 저장
final_csv_path = "/content/drive/MyDrive/졸프데이터/law_data_for_objectbox.csv"
df.to_csv(final_csv_path, index=False)

print(f"안드로이드용 준비 완료: {final_csv_path}")
```
